**ARRAY** **ATTRIBUTES**

```python
import numpy as np

a1=np.arange(10,dtype=np.int32)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
a3
```

```python
#ndim
a1.ndim
a2.ndim
a3.ndim
```

```python
#shape
print(a1.shape)
a1

print(a2.shape)
a2

print(a3.shape)
a3
```

```python
#size
print(a1.size)
print(a1)

print(a2.size)
print(a2)

print(a3.size)
print(a3)
```

```python
#itemsize -- tells every item is consuming how much space in memory
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
#dtype
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```
