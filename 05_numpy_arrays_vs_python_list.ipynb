**NUMPY** **ARRAY** **VS** **PYTHON** **LIST**

```python
#ON BASIS OF TIME
```

```python
#python list
import time
a=[i for i in range(10000000)]
b=[i for i in range(10000000,20000000)]
c=[]

start=time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time()-start)
```

```python
#numpy
import numpy as np
a=np.arange(10000000)
b=np.arange(10000000,20000000)
start=time.time()
c=a+b
print(time.time()-start)
```

```python
# ON BASIS OF MEMORY
```

```python
import sys
a=[i for i in range(10000000)]
sys.getsizeof(a)
```

```python
a1=np.arange(10000000,dtype=np.int16)
sys.getsizeof(a1)
```

**ADVANCED** **INDEXING**

```python
a=np.arange(24).reshape(6,4)
print(a)

print(" ")


#fancy indexing it is used when we are not able to find patterns

print(a[[0,3,4,5]])  #rowwise

print(" ")

print(a[:,[0,2,3]])  #columnwise
```

```python
#boolean indexing
a2=np.random.randint(1,100,24).reshape(6,4)
print(a2)


#find all the numbeer greater than 50
print(" ")
print(a2>50)

print(" ")
print(a2[a2>50])

print(" ")
print(a2[a2%2==0])

print(" ")
print(a2[(a2>50) & (a2%2==0)])
```
