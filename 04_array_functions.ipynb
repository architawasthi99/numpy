**ARRAY** **FUNCTIONS**

```python
import numpy as np
```

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
print(a1)
```

```python
print(np.max(a1))

print(np.min(a1))

print(np.sum(a1))

print(np.prod(a1))

print(np.max(a1,axis=1)) # row

print(np.max(a1,axis=0)) #column

print(np.sum(a1,axis=0))

print(np.prod(a1,axis=1))

print(np.sum(a1,axis=0))

print(" ")
print(np.mean(a1))
print(np.mean(a1,axis=1))
print(np.mean(a1,axis=0))
```

```python
print(np.sin(a1))

print(" ")

print(np.tan(a1))
```

```python
#DOT PRODUCT,log
a3=np.arange(12).reshape(3,4)
a4=np.arange(12,24).reshape(4,3)
print(a3)
print(" ")
print(a4)

print(" ")

print(np.dot(a3,a4))

print(" ")

print(np.log(a1))
```

**INDEXING** **AND** **SLICING**

```python
m1=np.arange(12)
m2=np.arange(12).reshape((3,4))
m3=np.arange(8).reshape(2,2,2)
m2
```

```python
print(m2[1,2])
print(m2[2,3])
```

```python
print(m3)

print(" ")

print(m3[1,0,1])
```

```python
#SLICING
print(m2)

print(" ")

print(m2[0,:])
print(m2[2,:])
print(m2[:,2])

print(" ")
print(m2[1:,1:3])

print(" ")
print(m2[0:,1:])

print(" ")
print(m2[::2,::3])

print(" ")
print(m2[0::2,1::2])
```

```python
m3=np.arange(27).reshape(3,3,3)
print(m3)
print(" ")
print(m3[1])

print(" ")
print(m3[0,1,:])
```

**IRERATING**

```python
print(m1)

for i in m1:
  print(i)
```

```python
print(m2)

for i in m2:
  print(i)
```

```python
print(m3)
print(" ")
for i in m3:
  print(i)

  print(" ")

for i in np.nditer(m3):
  print(i)
```

**STACKING**

```python
#HORIZONTAL STACKING
l1=np.arange(12).reshape(3,4)
l2=np.arange(12,24).reshape(3,4)

print(np.hstack((l1,l2)))

print(" ")

print(np.vstack((l1,l2)))
```
