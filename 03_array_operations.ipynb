**ARRAY** **OPERATIONS**

```python
import numpy as np
```

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1)
print(" ")
print(a2)
```

```python
#Arithmetic operations
print(a1+2)
print(" ")
print(a1*2)
print(" ")
print(a1%2)
```

```python
#relational operations
print(a1>15)

print(" ")

print(a1==53)

print(" ")

print(a1!=24)
```

```python
#Aggregate operations

print(np.sum(a1))

print(" ")

print(np.min(a1))
print(np.max(a1))
print(np.mean(a1))
print(np.median(a1))
print(np.var(a1))
print(np.exp(a1))
```

```python
#Searching

print(np.argmax(a2))

print(np.argmin(a2))
```

```python
#vector arithmetic operations

print (a1+a2)

print(" ")

print(a1*a2)

print(" ")

print(a1-a2)

print(" ")

print(a1**a2)
```
