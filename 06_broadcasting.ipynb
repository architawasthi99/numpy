**BROADCASTING**

```python
#same shape
import numpy as np

a1=np.arange(6).reshape(2,3)
a2=np.arange(6,12).reshape(2,3)

print(a1)
print(" ")
print(a2)
print(" ")
print(a1+a2)


print(" ")

a3=np.arange(3).reshape(1,3)
print(a1+a3)
```
