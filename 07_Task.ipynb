###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

###`Q-2:` Replace multiples of 3 or 5 as 0 in the given array.
```
arr=[1 2 3 4 5 6 7 9]

result-> [1 2 0 4 0 0 7 0]
```

````python
# code here
arr=np.array([1,2,3,4,5,6,7,9])
arr[(arr%3==0) | (arr%5==0)]=0
print(arr)

print(" ")

result=np.where((arr%3==0) | (arr%5==0),0,arr)
print(result)
````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here

arr = np.array([[1, 2, np.nan],
                [4, 2, 6],
                [np.nan, np.nan, 5]])


# first remove nan values
values=arr[~np.isnan(arr)]

# finding most common elements
unique,count=np.unique(values,return_counts=True)
most_common=unique[np.argmax(count)]
arr[np.isnan(arr)]=most_common
print(arr)
````

###`Q-6:` Given two arrays, X and Y, construct the Cauchy matrix C.
`Cij =1/(xi - yj)`

http://en.wikipedia.org/wiki/Cauchy_matrix
```
x = numpy.array([1,2,3,4]).reshape((-1, 1)
y = numpy.array([5,6,7])
```

````python
# code here
import numpy as np

x = np.array([1,2,3,4]).reshape((-1, 1))
y = np.array([5,6,7])
C = 1 / (x[:, np.newaxis] - y)

print(C)
````

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
# code here
import numpy as np
import matplotlib.pyplot as plt
def tanh(arr):
    return (np.exp(arr) - np.exp(-arr)) / (np.exp(arr) + np.exp(-arr))

arr = np.arange(-10, 11)
y = tanh(arr)

plt.plot(arr, y)
plt.xlabel("x")
plt.ylabel("tanh(x)")
plt.title("Tanh Function")
plt.grid()
plt.show()
````

### `Q-8:` Plot the below equation.

$$\large y = \sqrt{36 - (x - 4)^2} + 2$$

The range of x should be between -2 to 10. $x ∈ [-2, 10]$



````python
# code here

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 10, 500)
y = np.sqrt(36 - (x - 4)**2) + 2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
````
